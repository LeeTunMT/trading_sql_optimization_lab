## Validate Time Series Data

```python
import os
```

```python
def check_file(file_path):
    EXPECTED_COLUMNS = ['Date', 'Open', 'High', 'Low', 'Close', 'Adj Close', 'Volume']
    unexpected_files = []
    count = 0
    for dirname, _, filenames in os.walk(file_path):
        for filename in filenames:
            file_path = os.path.join(dirname, filename)
            df = pd.read_csv(file_path)
            if list(df.columns) != EXPECTED_COLUMNS:
                print(f"File {filename} has unexpected columns: {df.columns}")
                unexpected_files.append(filename)
                count += 1
    print(f"Total unexpected files: {count}")
    return unexpected_files
            
```

```python
unexpected_files = check_file('dataset/')
```

```python
print(unexpected_files)
```

```python
def clean_stock_df(df):
    df.columns = df.columns.str.lower().str.replace(' ', '_')
    df['date'] = pd.to_datetime(df['date'])
    df = df.dropna(subset=['date'])
    df = df.drop_duplicates(subset = ['date'])
    df = df[df['volume'] >= 0]
    df = df[df['low'] <= df['high']]
    return df
```
